```python
ASSIGNMENT 6
import pandas as pd
# Create sample dataset
data = {
    'Name': ['Aman', 'Riya', 'Karan', 'Riya', 'Neha'],
    'Marks': [85, None, 72, None, 90],
    'City': ['Pune', 'Mumbai', 'Delhi', 'Mumbai', 'Pune']
}
df = pd.DataFrame(data)
print("Original Data:")
print(df)

# Save sample data as CSV file
df.to_csv("student_data.csv", index=False)

# Read CSV file
df = pd.read_csv("student_data.csv")

print("\nFirst Five Rows:")
print(df.head())

# Display dataset information
print("\nDataset Information:")
print(df.info())

# Handle missing values using fillna()
df['Marks'] = df['Marks'].fillna(df['Marks'].mean())

print("\nAfter Handling Missing Values:")
print(df)

# Find duplicate records
print("\nDuplicate Records:")
print(df[df.duplicated()])

# Remove duplicate records
df = df.drop_duplicates()

# Remove irrelevant column (example)
# df = df.drop('City', axis=1)

# Filter students scoring more than 75 marks
filtered_df = df.loc[df['Marks'] > 75]

print("\nFiltered Data (Marks > 75):")
print(filtered_df)

# Save cleaned data
df.to_csv("cleaned_student_data.csv", index=False)

print("\nCleaned data saved successfully.")
```
